```python
##What are exceptions in python?
##exceptions are events that occur during the execution of a program that disrupt the normal flow of instructions.

#ZeroDivisionError: This exception occurs when you try to divide a number by zero. For example:
#zero_division = 10 / 0  # This will raise a ZeroDivisionError
#FileNotFoundError: This exception occurs when you try to access a file that does not exist. For example:
#ValueError: This exception occurs when you pass an argument of the correct type but an inappropriate value. For example:
#number = int("abc")  # This will raise a ValueError
#typeError: This exception occurs when you perform an operation on an object of an inappropriate type. For example:
#my_list = [1, 2, 3]
#my_list + 5  # This will raise a TypeError
```

```python
##Exception try,except block

try:
    a=b
except:
    print("The variavle has not been generated")
```

```python
try:
    a=b
except NameError as ex:
    print(f"NameError: {ex}")
```

```python
try:
    res=1/0
except ZeroDivisionError as ex:
    print(f"ZeroDivisionError: {ex}")
```

```python
##Main Exception Handling -- alwaays use the most specific exception first and then the more general ones.
try:
    result=1/0
    a=b
except ZeroDivisionError as ex:
    print(f"ZeroDivisionError: {ex}")
   
except Exception as ex:
    print(f"An error occurred: {ex}")
```

```python
#try, catch, else-- if try get executed successfully then else block will be executed otherwise except block will be executed.
try:
    a=0
except NameError as ex:
    print(f"NameError: {ex}")
else:
    print(f"The variable has been generated successfully: {a}")
```

```python
##try,catch,else,finally-- finally block will be executed no matter what happens in try and except block.
try:
    a=b
except NameError as ex:
    print(f"NameError: {ex}")
else:
    print(f"The variable has been generated successfully: {a}")
finally:
    print("This block will always be executed.")
```

```python
##File and handling and exception handling in python

try:
    file = open("new.txt", "r")
    content = file.read()
except FileNotFoundError as ex:
    print(f"FileNotFoundError: {ex}")
finally:
    if 'file' in locals() or not file.closed():
        file.close()
```
