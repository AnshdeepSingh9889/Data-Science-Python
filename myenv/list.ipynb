```python
to_do_list=['Buy Groceries','Clean the house','Pay bills']

##adding the task
to_do_list.append("Schedule Meeting")
to_do_list.append("Go for a run")


##Removing the task
to_do_list.remove("Clean the house")

##checkig if a task is in the list
if 'Pay bills' in to_do_list:
    print("Don't forget to pay the utility bills")
print(to_do_list) 

print("To do List remainig")
for task in to_do_list:
    print(f"{task}")
```

```python
##Organizing the Grades of student

grades=[85,92,78,99,88]

grades.append(95)

average_grade=sum(grades)/len(grades)

print(f"average grade:{average_grade:.7f}")

#finding the highest and lowest grades

highest_grade=max(grades)
lowest_grade=min(grades)
print(f"Highest Grade:{highest_grade:.2f} ")
print(f"Lowest Grade: {lowest_grade:.3f}")

```

```python
##feedback
feedback=['Great service','Satisfied','do better','Excellent expereience']
feedback.append("Not happy with the service")

positive_feedback=sum(1 for comment in feedback if 'great' in comment.lower() or 'excellent' in comment.lower())

print(f"positive feedback :{positive_feedback}")

#print all the feedback

for i in feedback:
    print(f"{i}")
```
