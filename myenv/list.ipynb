```python
to_do_list=['Buy Groceries','Clean the house','Pay bills']

##adding the task
to_do_list.append("Schedule Meeting")
to_do_list.append("Go for a run")


##Removing the task
to_do_list.remove("Clean the house")

##checkig if a task is in the list
if 'Pay bills' in to_do_list:
    print("Don't forget to pay the utility bills")
print(to_do_list) 

print("To do List remainig")
for task in to_do_list:
    print(f"{task}")
```

```python
##Organizing the Grades of student

grades=[85,92,78,99,88]

grades.append(95)

average_grade=sum(grades)/len(grades)

print(f"average grade:{average_grade:.7f}")

#finding the highest and lowest grades

highest_grade=max(grades)
lowest_grade=min(grades)
print(f"Highest Grade:{highest_grade:.2f} ")
print(f"Lowest Grade: {lowest_grade:.3f}")

```

```python
##feedback
feedback=['Great service','Satisfied','do better','Excellent expereience']
feedback.append("Not happy with the service")

positive_feedback=sum(1 for comment in feedback if 'great' in comment.lower() or 'excellent' in comment.lower())

print(f"positive feedback :{positive_feedback}")

#print all the feedback

for i in feedback:
    print(f"{i}")
```

```python


##Writing and then reading a file

with open('new.txt','w+') as dile:
    dile.write("Hello world\n")
    dile.write("This is a new line\n")
    # move the file cursor to 1st one to read

    dile.seek(0)

    ##read the content of the file
    red=dile.read()
    print(red)
```

```python
##read the lines, words and characters in a text file:

with open('new1.txt','r') as file:
    for line in file:
        words=line.split()
        for word in words:
            for i in word:
                print(i,end=' ')
            print(word,end=' ')
```

```python
## read the content from a file as paste in new file

with open('new.txt','r') as file, open('new1.txt','w') as dest:
    dest.write(file.read())
```

```python
path='new.txt'
if os.path.exists(path):
    print(f"The file {path} exists.")
else:
    print(f"The file {path} does not exist.")
```

```python
##checking if a path is a file or directory
import os

path='new.txt'

if os.path.isfile(path):
    print(f"{path} is a file.")
elif os.path.isdir(path):
    print(f"{path} is a directory.")
else:
    print(f"{path} does not exist.")
```

```python
##getting absolute path of a file or directory
import os

path='new.txt'
absolute_path=os.path.abspath(path)
print(absolute_path)
```
