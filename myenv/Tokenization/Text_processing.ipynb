```python
text='Learning is a continuous process. It is important to keep learning new things every day. The more you learn,the more you grow. Learning can be fun and exciting if you approach it with the right mindset.'
tokens=word_tokenize(text)
print("Tokens :",tokens)



```

```python
#sentence tokenization

sentences = sent_tokenize(text)
print("Sentences :",sentences)
```

```python
#Removing stopwords
stop_words=set(stopwords.words('english'))
filtered_tokens=[token for token in  tokens if token.lower() not in stop_words]
print("Filtered Tokens :",filtered_tokens)
```

```python
# Stemming: reducing words to their root form using Porter Stemmer
porter_stemmer = PorterStemmer()
stemmed_tokens = [porter_stemmer.stem(token) for token in filtered_tokens]
print("Stemmed Tokens :",stemmed_tokens)
```

```python
#Lemmatization: reducing words to their base form using WordNet Lemmatizer
wordnet_lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [wordnet_lemmatizer.lemmatize(token) for token in filtered_tokens]
print("Lemmatized Tokens :",lemmatized_tokens)
```

```python
# Special character removal
special_characters = string.punctuation
filtered_tokens_no_special = [token for token in filtered_tokens if token not in special_characters]
print("Filtered Tokens without Special Characters :",filtered_tokens_no_special)
```
