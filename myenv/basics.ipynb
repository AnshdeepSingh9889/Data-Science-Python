```python
#list comprehension syntax 
Basically it is [expression for item in iterable if condition]
```

```python
square=[x**2 for x in range (20)]
print(square)
```

```python
SQUARE=[    x**2 for x in range(20) if x%2==0]
print(SQUARE)
```

```python
##nested list comprehension-> it is a list comprehension inside another list comprehension
from numpy import iterable


[expression for item1  in iterable for item2 in iterable2 if condition]
```

```python
lst1=[1,2,3,4,5 ]
lst2=['a','b','c','d','e']
pair=[[i,j] for i in lst1 for j in lst2]
print(pair)
```

```python
##list comprehension with function call
word_list=['hello','world','python','programming']
length_list=[len(word) for word in word_list]
print(length_list)
```
