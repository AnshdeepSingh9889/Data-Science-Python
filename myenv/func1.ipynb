```python
##csv 

import csv

with open('example.csv',mode='w',newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name','age'])
    
    writer.writerow(['krish',32])

with open('example.csv',mode='r')as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
##Read whole file

with open("new.txt",mode='r') as file:
    fileread=file.read()
    print(fileread)
```

```python
##read line by line
with open("new.txt",'r') as file:
    for linr in file:
        print(linr.strip())## strip removes the new line character
```

```python
##write a file overwriting

with open('new.txt','w') as file:
    file.write("Hello world anshdeep\n")
    file.write('this is the new line')
```

```python
## write a file in append mode

with open('new.txt','a') as file:
    file.write("Append the text in the given files\n")
```

```python
##writing a list of lines to a file

list=['first line\n','second line \n','third line\n']

with open("new.txt",'a') as file:
    for i in list:
        file.write(i)

        
```
