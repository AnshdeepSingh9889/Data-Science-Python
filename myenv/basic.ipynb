```python
##Loop Control statements
##break -- exits the break prematurely

for i in range(10):
    if(i==5):
        break
    print(i)
```

```python
##continue-- COntinue skips the iteration and moves to next condition

for  i in range(10):
    if(i%2==0):
        continue
    print(i)
```

```python
## pass- it is the null operation and does nothing

for i in range(5):
    if i==3:
        pass
    print(i)
```
