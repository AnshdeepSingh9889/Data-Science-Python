```python
##variable length arguments
##positional and keyword argument *Args

def print_numbers(*a):
    for numbers in a:
        print(numbers)
```

```python
print_numbers('a',3,4,45,44,5,5,67,7,4,3,2,2,2,43,5)
```

```python
##keyword arguments : all the argument in key value pair

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="ANsh",age='32',country='India')
```

```python
def print_det(*args,**kwargs):
    for val in args:
        print(f"positional argument: {val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_det(1,2,3,4,"ansh", name="Ansh", age="32",country="India")
```
