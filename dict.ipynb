```python
student={'Name':"Ansh",'age':'32','class':'Btech'}

print(student)
```

```python
##Nested dictionaries
students={
    "student1":{'name':'Ansh','age':20},
    "Sudent2":{'name':'deep','age':21}
}
print(students)
```

```python
#Acess the nested dictionary
print(students['student1']['name'])
print(students['Sudent2']['age'])
```

```python
students.items()
```

```python
##iterating over nested dictioanry
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")   
    for key,value in student_info.items():
        print(f'{key}:{value}')
```

```python
##Dictionay comprehension

squares={x:x**2 for x in range(5)}
print(squares)
```

```python
cubes={x:x**3 for x in range (10) if x%2==0}
print(cubes)
```

```python
##use the dictionary to find the frequency of element in the list

numbers=[1,2,2,2,2,2,3,3,3,3,3,1,1,1,3,4,5,5,6,6]
frequency={}
for num in  numbers:
    if num in frequency:
        frequency[num]+=1
    else:
        frequency[num]=1
print(frequency)
```
